Notebook นี้สรุปแนวคิด Big-O พร้อมตัวอย่างโค้ด Python และคำอธิบายทีละขั้น เหมาะสำหรับใช้สอนในรายวิชา Data Structures for Optimal Solutions โดยเชื่อมกับหัวข้อการค้นหา การเรียงลำดับ และการเลือกอัลกอริทึมให้เหมาะกับปัญหา

## จุดประสงค์การเรียนรู้

เมื่อเรียนจบหัวข้อนี้ นักศึกษาควรจะสามารถอธิบายได้ว่า Big-O คืออะไร, ใช้เพื่อเปรียบเทียบอัลกอริทึมอย่างไร, อ่านความซับซ้อนจากโค้ดเบื้องต้นได้, และอธิบายความแตกต่างระหว่างอัลกอริทึมอย่าง Linear Search, Binary Search, Bubble Sort และ Merge Sort ได้

## 1) Big-O คืออะไร

Big-O notation เป็นวิธีอธิบายอัตราการเติบโตของเวลาในการทำงาน หรือการใช้หน่วยความจำของอัลกอริทึม เมื่อขนาดข้อมูลอินพุต `n` เพิ่มขึ้นมาก ๆ

แนวคิดสำคัญคือ เราไม่ได้สนใจเวลาจริงเป็นวินาที เพราะเวลาจริงขึ้นกับเครื่อง ภาษา และการติดตั้งระบบ แต่เราสนใจว่าเมื่อข้อมูลโตขึ้น อัลกอริทึมจะช้าลงเร็วแค่ไหน

ตัวอย่างเช่น

- ถ้าอัลกอริทึมใช้เวลาเพิ่มตามจำนวนข้อมูลแบบเส้นตรง เราเขียนว่า `O(n)`
- ถ้าเวลาโตแบบกำลังสอง เราเขียนว่า `O(n^2)`
- ถ้าเวลาค่อย ๆ โตตามการหารครึ่งซ้ำ ๆ เราเขียนว่า `O(log n)`

Big-O จึงเป็นเครื่องมือสำหรับเปรียบเทียบความ scalable ของวิธีแก้ปัญหาหลายแบบ

## 2) ทำไมต้องเรียน Big-O

ในวิชาโครงสร้างข้อมูล เราไม่ได้ต้องการแค่ให้โปรแกรมทำงานได้ แต่ต้องการให้เลือกวิธีที่เหมาะสมกับขนาดข้อมูลด้วย

ถ้าข้อมูลมีแค่ 10 ตัว วิธีไหนก็อาจเร็วพอ แต่ถ้าข้อมูลมี 10,000 หรือ 1,000,000 ตัว ความแตกต่างระหว่าง `O(n)`, `O(log n)` และ `O(n^2)` จะชัดเจนมาก

ดังนั้น Big-O ช่วยให้เรา

- เปรียบเทียบอัลกอริทึมอย่างเป็นระบบ
- เลือกโครงสร้างข้อมูลให้เหมาะกับงาน
- ประเมินได้ว่าโปรแกรมจะยังใช้ได้ดีหรือไม่เมื่อข้อมูลเพิ่มขึ้น

## 3) ลำดับการเติบโตที่พบบ่อย

จากเร็วไปช้ากว่าโดยทั่วไป

1. `O(1)` คงที่
2. `O(log n)` ลอการิทึม
3. `O(n)` เชิงเส้น
4. `O(n log n)`
5. `O(n^2)` กำลังสอง
6. `O(n^3)` กำลังสาม
7. `O(2^n)` เอ็กซ์โปเนนเชียล
8. `O(n!)` แฟกทอเรียล

ยิ่งอยู่ล่างรายการมาก เมื่อ `n` โตขึ้น เวลาก็ยิ่งพุ่งเร็ว

```python
growth_examples = {
    'O(1)': lambda n: 1,
    'O(log n)': lambda n: 0 if n <= 0 else __import__('math').log2(n),
    'O(n)': lambda n: n,
    'O(n log n)': lambda n: 0 if n <= 0 else n * __import__('math').log2(n),
    'O(n^2)': lambda n: n**2,
}

sizes = [10, 100, 1000, 10000]
for name, fn in growth_examples.items():
    print(name)
    for n in sizes:
        print(f'  n={n:5d} -> {fn(n):,.2f}')
    print()
```

## 4) ความหมายของแต่ละแบบแบบสั้น ๆ

### `O(1)`
ใช้เวลาเท่าเดิมไม่ว่าข้อมูลจะมีขนาดเท่าไร เช่น การเข้าถึงสมาชิกใน list ด้วย index หนึ่งตำแหน่ง

### `O(log n)`
มักเกิดเมื่อปัญหาถูกลดลงครึ่งหนึ่งทุกครั้ง เช่น Binary Search

### `O(n)`
ต้องไล่ดูข้อมูลทีละตัว เช่น Linear Search

### `O(n log n)`
พบบ่อยในอัลกอริทึมเรียงลำดับที่มีประสิทธิภาพ เช่น Merge Sort

### `O(n^2)`
มักเกิดเมื่อมีลูปซ้อนกันสองชั้นที่ทำงานกับข้อมูลชุดเดียวกัน เช่น Bubble Sort

## 5) กฎพื้นฐานในการหา Big-O จากโค้ด

### กฎที่ 1: ค่าคงที่ตัดทิ้งได้
ถ้ามีการทำงาน 5 ครั้ง, 100 ครั้ง หรือ 1000 ครั้ง แต่ไม่ขึ้นกับ `n` ก็ยังถือเป็น `O(1)`

### กฎที่ 2: เก็บเฉพาะพจน์ที่โตเร็วที่สุด
เช่น `O(n^2 + n + 1)` สรุปเป็น `O(n^2)`

### กฎที่ 3: ลูปหนึ่งชั้นมักเป็น `O(n)`
ถ้าลูปวิ่งครบข้อมูล `n` ตัวครั้งเดียว ก็เป็นเชิงเส้น

### กฎที่ 4: ลูปซ้อนสองชั้นมักเป็น `O(n^2)`
ถ้าทั้งลูปนอกและลูปในโตตาม `n`

### กฎที่ 5: ถ้าหารครึ่งซ้ำ ๆ มักเป็น `O(log n)`
เช่น `n = n // 2`

### กฎที่ 6: ถ้ามีขั้นตอนต่อกัน ให้บวก
เช่น `O(n)` แล้วตามด้วย `O(n^2)` จะรวมเป็น `O(n + n^2)` และย่อเป็น `O(n^2)`

```python
def example_constant(arr):
    return arr[0]


def example_linear(arr):
    total = 0
    for x in arr:
        total += x
    return total


def example_quadratic(arr):
    count = 0
    for i in arr:
        for j in arr:
            count += 1
    return count
```

## 6) ตัวอย่างที่ 1: Linear Search

แนวคิดของ Linear Search คือไล่ดูข้อมูลจากซ้ายไปขวาทีละตัว จนกว่าจะเจอค่าที่ต้องการ หรือจนกว่าจะครบทุกตัว

ถ้าข้อมูลไม่เรียงลำดับมาก่อน วิธีนี้ใช้ง่ายและตรงไปตรงมา แต่ในกรณีแย่ที่สุดอาจต้องตรวจครบ `n` ตัว จึงมีเวลาเป็น `O(n)`

```python
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

numbers = [4, 8, 15, 16, 23, 42]
print('index of 15 =', linear_search(numbers, 15))
print('index of 99 =', linear_search(numbers, 99))
```

### วิเคราะห์เวลา

- กรณีดีที่สุด: เจอที่ตัวแรก ใช้เวลา `O(1)`
- กรณีเฉลี่ย: ตรวจหลายตัว ใช้เวลาประมาณ `O(n)`
- กรณีแย่ที่สุด: เจอที่ตัวสุดท้ายหรือไม่เจอเลย ใช้เวลา `O(n)`

ดังนั้นเวลาที่ใช้อธิบายโดยรวมของ Linear Search มักเขียนเป็น `O(n)`

## 7) ตัวอย่างที่ 2: Binary Search

Binary Search ใช้ได้เมื่อข้อมูลเรียงลำดับแล้วเท่านั้น หลักการคือดูค่าตรงกลาง ถ้าค่าน้อยไปก็ทิ้งครึ่งซ้าย ถ้าค่ามากไปก็ทิ้งครึ่งขวา ทำแบบนี้ซ้ำไปเรื่อย ๆ

เพราะขนาดปัญหาถูกลดลงครึ่งหนึ่งทุกครั้ง จึงมีเวลาเป็น `O(log n)` ซึ่งเร็วกว่า Linear Search มากเมื่อข้อมูลมีขนาดใหญ่

```python
def binary_search(arr, target):
    left = 0
    right = len(arr) - 1

    while left <= right:
        mid = (left + right) // 2

        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return -1

sorted_numbers = [3, 7, 11, 18, 24, 31, 45, 59, 62]
print('index of 24 =', binary_search(sorted_numbers, 24))
print('index of 50 =', binary_search(sorted_numbers, 50))
```

### วิเคราะห์เวลา

ถ้ามีข้อมูล 16 ตัว การหารครึ่งจะเป็นประมาณ 16 -> 8 -> 4 -> 2 -> 1 จึงใช้ไม่กี่รอบ

ถ้ามีข้อมูล 1,024 ตัว ก็ยังใช้เพียงประมาณ 10 รอบเท่านั้น เพราะ `log2(1024) = 10`

ดังนั้น Binary Search จึงเหมาะมากเมื่อ

- ข้อมูลถูกเรียงไว้แล้ว
- มีการค้นหาบ่อย
- ข้อมูลมีขนาดใหญ่

## 8) เปรียบเทียบ Linear Search กับ Binary Search

คำถามสำคัญในชั้นเรียนคือ ทำไม Binary Search ถึงเร็วกว่า คำตอบคือ Linear Search ตัดปัญหาออกทีละ 1 ตัวโดยประมาณ แต่ Binary Search ตัดออกทีละครึ่งหนึ่ง

อย่างไรก็ตาม Binary Search ไม่ได้ดีกว่าเสมอไป เพราะถ้าข้อมูลยังไม่เรียง การเตรียมข้อมูลอาจมีต้นทุนเพิ่มขึ้น

```python
import math

sizes = [10, 100, 1000, 1_000_000]
print(f"{'n':>12} {'linear ~ n':>15} {'binary ~ log2(n)':>20}")
for n in sizes:
    print(f"{n:12,d} {n:15,.0f} {math.log2(n):20.2f}")
```

## 9) ตัวอย่างที่ 3: Bubble Sort

Bubble Sort เป็นอัลกอริทึมเรียงลำดับพื้นฐานที่เปรียบเทียบสมาชิกที่อยู่ติดกัน แล้วสลับตำแหน่งถ้าลำดับผิด ทำซ้ำหลายรอบจนข้อมูลเรียงเสร็จ

ข้อดีคือแนวคิดง่ายและเหมาะใช้สอนการทำงานของการสลับค่า แต่ข้อเสียคือเมื่อข้อมูลโตขึ้นจะช้าค่อนข้างมาก

```python
def bubble_sort(arr):
    a = arr.copy()
    n = len(a)

    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swapped = True
        if not swapped:
            break
    return a

sample = [5, 1, 4, 2, 8]
print('before:', sample)
print('after :', bubble_sort(sample))
```

### วิเคราะห์เวลา

Bubble Sort มีลูปซ้อนกันเป็นหลัก จึงมักมีเวลาในกรณีเฉลี่ยและกรณีแย่ที่สุดเป็น `O(n^2)`

- กรณีดีที่สุด: ถ้าข้อมูลเรียงอยู่แล้ว และเราใส่ตัวแปร `swapped` เพื่อตรวจสอบ อาจใกล้ `O(n)`
- กรณีเฉลี่ย: `O(n^2)`
- กรณีแย่ที่สุด: `O(n^2)`

ดังนั้น Bubble Sort เหมาะกับการเรียนแนวคิด แต่ไม่เหมาะกับข้อมูลขนาดใหญ่

## 10) ตัวอย่างที่ 4: Merge Sort

Merge Sort ใช้แนวคิด divide and conquer คือ

1. แบ่งข้อมูลออกเป็นครึ่งซ้ายและครึ่งขวา
2. เรียงแต่ละครึ่งแบบ recursive
3. นำผลลัพธ์ที่เรียงแล้วมารวมกันอย่างเป็นระเบียบ

เพราะแต่ละระดับของการแบ่งใช้เวลารวมประมาณ `O(n)` และจำนวนระดับมีประมาณ `O(log n)` จึงรวมเป็น `O(n log n)`

```python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    return merge(left, right)


def merge(left, right):
    result = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])
    return result

sample = [38, 27, 43, 3, 9, 82, 10]
print('before:', sample)
print('after :', merge_sort(sample))
```

### วิเคราะห์เวลา

Merge Sort ใช้เวลา `O(n log n)` ทั้งในกรณีเฉลี่ยและกรณีแย่ที่สุด จึงมีประสิทธิภาพดีกว่า Bubble Sort มากเมื่อข้อมูลมีขนาดใหญ่

อย่างไรก็ตาม Merge Sort มักใช้หน่วยความจำเพิ่มเติมในการ merge ข้อมูล จึงเป็นตัวอย่างที่ดีว่าในการเลือกอัลกอริทึม เราไม่ได้ดูแค่เวลา แต่ควรดูพื้นที่หน่วยความจำด้วย

## 11) เปรียบเทียบ Bubble Sort กับ Merge Sort

Bubble Sort เข้าใจง่าย แต่โตเป็น `O(n^2)` ส่วน Merge Sort ซับซ้อนกว่าเล็กน้อย แต่มีเวลาเป็น `O(n log n)` จึงเหมาะกับข้อมูลขนาดใหญ่กว่าอย่างชัดเจน

```python
sizes = [10, 100, 1000, 10000]
print(f"{'n':>10} {'n^2':>15} {'n log2 n':>20}")
for n in sizes:
    print(f"{n:10,d} {n**2:15,.0f} {n * math.log2(n):20,.2f}")
```

## 12) Time Complexity กับ Space Complexity

Big-O ไม่ได้ใช้กับเวลาอย่างเดียว แต่ใช้กับพื้นที่หน่วยความจำได้ด้วย

- Time Complexity คือเวลาในการทำงานเมื่อ `n` โตขึ้น
- Space Complexity คือหน่วยความจำเพิ่มเติมที่ใช้งานเมื่อ `n` โตขึ้น

ตัวอย่างเช่น

- Linear Search ใช้พื้นที่เพิ่มน้อยมาก
- Merge Sort เร็วกว่า Bubble Sort ในหลายกรณี แต่ใช้พื้นที่ช่วยในการ merge มากกว่า

ดังนั้นในการสอนควรย้ำว่า เร็วกว่า ไม่ได้แปลว่า ดีกว่าเสมอ ถ้ามีข้อจำกัดด้านหน่วยความจำ

## 13) วิธีอ่าน Big-O จากโค้ดอย่างรวดเร็ว

### กรณีที่ 1: ลูปเดียว
ถ้าลูปวิ่งจาก `0` ถึง `n-1` หนึ่งรอบ มักเป็น `O(n)`

### กรณีที่ 2: ลูปซ้อน
ถ้าทุกลูปโตตาม `n` มักเป็น `O(n^2)` หรือสูงกว่า

### กรณีที่ 3: ลดครึ่งทุกครั้ง
ถ้าใช้ `while n > 1:` แล้วทำ `n = n // 2` มักเป็น `O(log n)`

### กรณีที่ 4: recursion แบบแบ่งครึ่งแล้วรวมผล
มักนำไปสู่ `O(n log n)` เช่น Merge Sort

```python
def log_example(n):
    steps = 0
    while n > 1:
        n //= 2
        steps += 1
    return steps

for n in [8, 16, 32, 64, 128]:
    print(f'n={n:3d} -> steps={log_example(n)}')
```

## 14) ข้อควรระวัง Big-O

มักสับสนเรื่องต่อไปนี้

1. Big-O ไม่ใช่เวลาจริงเป็นวินาที แต่เป็นแนวโน้มการเติบโต
2. Big-O ไม่ได้บอกทุกอย่าง เพราะค่าคงที่จริงและรายละเอียดการ implement ก็มีผล
3. อัลกอริทึมที่ Big-O ดีกว่า อาจไม่ชนะเสมอในข้อมูลเล็ก
4. ต้องแยก best, average, worst case ออกจากกัน
5. บางอัลกอริทึมมี trade-off ระหว่างเวลาและพื้นที่

| อัลกอริทึม | ใช้ทำอะไร | เงื่อนไขสำคัญ | Time Complexity โดยย่อ |

## 17) โจทย์ทดลองเพิ่มเติม

ส่วนนี้ช่วยให้ใช้ notebook ในห้องเรียนได้จริง โดยให้นักศึกษาลองแก้ไขข้อมูลเอง แล้วรันใหม่เพื่อสังเกตผล

```python
test_data = [12, 7, 19, 3, 25, 1, 8]
print('original      :', test_data)
print('linear search :', linear_search(test_data, 25))
print('bubble sort   :', bubble_sort(test_data))

sorted_data = sorted(test_data)
print('sorted data   :', sorted_data)
print('binary search :', binary_search(sorted_data, 25))
print('merge sort    :', merge_sort(test_data))
```
